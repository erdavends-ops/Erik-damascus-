```python
volume = [35, 42, 48, 45]
total = sum(volume)
rata_rata = total / len(volume)
tertinggi = max(volume)
terendah = min(volume)

print("Total:", total)
print("Rata-rata:", rata_rata)
print("Tertinggi:", tertinggi)
print("Terendah:", terendah)
```

```python
import numpy as uhuy

volume = uhuy.array([35, 42, 48, 45])
print("Kabeh:", uhuy.sum(volume))
print("Roto-roto:", uhuy.mean(volume))
print("Paling gede:", uhuy.max(volume))
print("Cilik dewe:", uhuy.min(volume))
```

```python
import numpy as np

panjang = uhuy.array([4.0, 5.0, 6.0]) # meter
lebar = uhuy.array([3.0, 3.0, 2.5])   # meter
tebal = 0.12                          # meter

luas = panjang * lebar
volume = luas * tebal

print("Ombone Plat:", luas, "m2")
print("Volume isine pelat:", volume, "m3")
print("Jumlahnipun volume:", uhuy.sum(volume), "m3")
```
